```python
import json
import requests
from datetime import datetime, timezone

URL = "https://gbfs.citibikenyc.com/gbfs/en/station_status.json"

resp = requests.get(URL, timeout=30)   # give up if the site hangs
resp.raise_for_status()                # fail loudly on 404/500 instead of saving an error page
data = resp.json()

ts = datetime.now(timezone.utc).strftime("%Y%m%d_%H%M%S")
path = f"/Volumes/citibike/bronze/landing/station_status/{ts}.json"

with open(path, "w") as f:
    json.dump(data, f)

print(f"Saved {len(data['data']['stations'])} stations to {path}")
```
